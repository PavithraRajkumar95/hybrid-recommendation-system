```python
import pandas as pd
```

```python
movies=pd.read_csv(r"C:\Users\Dell\Documents\My ML Projects\Hybrid Recommendation System\data\movie.csv")
ratings=pd.read_csv(r"C:\Users\Dell\Documents\My ML Projects\Hybrid Recommendation System\data\rating.csv")
```

```python
#take subset of users
sample_users = ratings['userId'].sample(5000,random_state=42)
ratings_sample=ratings[ratings['userId'].isin(sample_users)]

#Merge
df = ratings_sample.merge(movies, on = 'movieId')

df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
df.shape
```

**Preprocessing**

Keeping only useful columns

```python
df = df[['userId', 'movieId', 'rating', 'title', 'genres']]
```

```python
df.columns
```

Remove Duplicates

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

Handling genres

```python
df['genres'] = df['genres'].str.replace('|',' ')
```

```python
df.head()
```

Handle very rare movies

```python
movie_counts = df['title'].value_counts()

df = df[df['title'].isin(movie_counts[movie_counts > 10].index)]
```

```python
df.info()
```

Remove duplicate movies for content based

```python
movies_df = df[['movieId', 'title', 'genres']].drop_duplicates()
```

```python
movies_df.shape
```

reset index

```python
movies_df = movies_df.reset_index(drop=True)
```

```python
movies_df.shape
```

Saving processed data

```python
df.to_csv("C:/Users/Dell/Documents/My ML Projects/Hybrid Recommendation System/data/processed_ratings.csv", index=False)
```
