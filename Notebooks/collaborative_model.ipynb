**Reducing the data set**


```python
top_movies = df['title'].value_counts().head(2000).index
df = df[df['title'].isin(top_movies)]
```

**Create User Item Matrix**

```python
user_item_matrix = df.pivot_table(
    index='userId',
    columns='title',
    values='rating'
)
```

fill missing values

```python
user_item_matrix = user_item_matrix.fillna(0)
```

**Compute Similarity**

```python
from sklearn.metrics.pairwise import cosine_similarity

item_similarity= cosine_similarity(user_item_matrix.T)
```

**Convert to DataFrame**

```python
item_similarity_df = pd.DataFrame(
    item_similarity,
    index=user_item_matrix.columns,
    columns=user_item_matrix.columns
)
```

Recommendation Function


```python
def recommend_cf(movie_name, n=10):

    similar_scores = item_similarity_df[movie_name]

    similar_scores = similar_scores.sort_values(ascending=False)

    return similar_scores[1:n+1]
```

Test

```python
recommend_cf("Toy Story (1995)")
```
